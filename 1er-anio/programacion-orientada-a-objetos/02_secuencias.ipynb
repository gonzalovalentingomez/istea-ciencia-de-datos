## Ejercicios

### 🟢 Ejercicio 1 — Manipulación de strings
Dado `"  Hola Mundo Python  "`, quitar espacios, convertir a mayúsculas y reemplazar "MUNDO" con "CLASE".


```python
# 🎯 EJERCICIO 1: Manipulación de strings
cadena = "  Hola Mundo Python  "
resultado = cadena.strip().upper().replace("MUNDO", "CLASE")
print(repr(resultado))
```

### 🟢 Ejercicio 2 — Operaciones con listas
Crear una lista con 5 ciudades, ordenarla alfabéticamente, invertirla, e imprimir la primera y la última.


```python
# 🎯 EJERCICIO 2: Operaciones con listas de ciudades
ciudades = ["Rosario", "Zárate", "Córdoba", "Mendoza", "Salta"]

ciudades.sort()
print(f"Ordenadas:  {ciudades}")

ciudades.reverse()
print(f"Invertidas: {ciudades}")

print(f"Primera: {ciudades[0]} | Última: {ciudades[-1]}")
```

### 🟡 Ejercicio 3 — Contador de vocales
Contar cuántas veces aparece cada vocal (a, e, i, o, u) en una oración dada.


```python
# 🎯 EJERCICIO 3: Contar vocales en una oración
oracion = "La programación en Python es apasionante y divertida"
vocales = "aeiou"

# Paso a minúsculas y saco las tildes para que "ó" cuente como "o"
texto = oracion.lower().translate(str.maketrans("áéíóú", "aeiou"))

conteo = {vocal: texto.count(vocal) for vocal in vocales}
print(", ".join(f"{vocal}: {cantidad}" for vocal, cantidad in conteo.items()))
```

### 🟡 Ejercicio 4 — Unpacking de nombre completo
De "Ana María García López", extraer el primer nombre y el último apellido usando split.


```python
# 🎯 EJERCICIO 4: Extraer primer nombre y último apellido
nombre_completo = "Ana María García López"

# Unpacking: el * se queda con todo lo del medio
primer_nombre, *_, ultimo_apellido = nombre_completo.split()

print(f"Primer nombre: {primer_nombre!r}")
print(f"Último apellido: {ultimo_apellido!r}")
```

### 🔴 Ejercicio 5 — List comprehension con filtro
Dada una lista de números, crear una nueva lista con solo los pares usando list comprehension.


```python
# 🎯 EJERCICIO 5: Filtrar pares con list comprehension
numeros = [3, 14, 7, 22, 9, 18, 5, 30, 11, 26]

pares = [n for n in numeros if n % 2 == 0]
print(pares)
```

## Desafío extra 🔴

Escribir una función que analice un texto largo (párrafo) y retorne un dict con:
- Las 5 palabras más frecuentes (ignorando palabras de menos de 4 letras)
- La longitud promedio de las palabras
- El porcentaje del texto que representan las 5 palabras más frecuentes

Probar con el fragmento del Martín Fierro incluido.


```python
# 🔴 Desafío: Análisis de texto
import re
from collections import Counter

texto = (
    "Aquí me pongo a cantar al compás de la vigüela,\n"
    "que el hombre que lo desvela una pena estraordinaria\n"
    "como la ave solitaria con el cantar se consuela.\n"
    "Pido a los santos del cielo que ayuden mi pensamiento,\n"
    "les pido en este momento que voy a cantar mi historia\n"
    "me refresquen la memoria y aclaren mi entendimiento.\n"
)

def analizar_texto(texto, top=5, largo_minimo=4):
    palabras = re.findall(r"[a-záéíóúüñ]+", texto.lower())
    largas = [p for p in palabras if len(p) >= largo_minimo]
    mas_frecuentes = Counter(largas).most_common(top)
    apariciones_top = sum(cantidad for _, cantidad in mas_frecuentes)
    return {
        "top_5_palabras": mas_frecuentes,
        "longitud_promedio": round(sum(len(p) for p in palabras) / len(palabras), 2),
        "porcentaje_del_texto": round(apariciones_top / len(palabras) * 100, 1),
    }

for clave, valor in analizar_texto(texto).items():
    print(f"{clave}: {valor}")
```

## Resumen de conceptos

| Concepto | Ejemplo | Descripción |
|---|---|---|
| str.strip() | `" hola ".strip()` | Elimina espacios extremos |
| str.split() | `"a,b".split(",")` | Divide en lista |
| str.join() | `",".join(["a","b"])` | Une lista en string |
| str.replace() | `"ab".replace("a","x")` | Reemplaza ocurrencias |
| f-string | `f"hola {nombre}"` | Interpolación de variables |
| list.append() | `lst.append(x)` | Agrega al final |
| list.sort() | `lst.sort()` | Ordena in-place |
| sorted() | `sorted(lst)` | Retorna nueva lista ordenada |
| Slicing | `lst[1:4]` | Sub-secuencia |
| Unpacking | `a, b = (1, 2)` | Asignación múltiple |
| List comprehension | `[x**2 for x in lst]` | Lista derivada compacta |
