## Análisis de Regresión Lineal

La regresión lineal modela la relación entre una variable $y$ y una variable $x$ con una recta.

**Fórmula:**

$$\hat{y} = \beta_0 + \beta_1 x$$

```python
import numpy as np
n = len(x)
beta_1 = (n*np.sum(x*y) - np.sum(x)*np.sum(y)) / (n*np.sum(x**2) - np.sum(x)**2)
beta_0 = np.mean(y) - beta_1 * np.mean(x)
```

# `busqueda_binaria(lista, objetivo)`

![versión](https://img.shields.io/badge/versión-1.0-blue) ![python](https://img.shields.io/badge/python-3.10%2B-green)

Busca un elemento en una lista **ordenada** dividiendo el espacio de búsqueda a la mitad en cada paso.
Como en cada comparación se descarta la mitad de los elementos, la cantidad máxima de comparaciones es:

$$\text{comparaciones} \le \lceil \log_2(n + 1) \rceil$$

Para una lista de 1.000.000 de elementos, alcanza con **20 comparaciones** como máximo.

## Parámetros

| Parámetro | Tipo | Descripción |
|:---|:---:|:---|
| `lista` | `list` | Lista ordenada de menor a mayor |
| `objetivo` | `int`, `float`, `str` | Elemento a buscar (comparable con los de la lista) |

## Retorna

| Tipo | Descripción |
|:---:|:---|
| `int` | Índice del elemento si se encuentra |
| `int` | `-1` si el elemento no está en la lista |

## Ejemplos de uso

```python
numeros = list(range(0, 100, 2))        # [0, 2, 4, ..., 98]
busqueda_binaria(numeros, 42)           # → 21
busqueda_binaria(numeros, 43)           # → -1

nombres = ["Ana", "Bruno", "Carla", "Diego"]
busqueda_binaria(nombres, "Carla")      # → 2
```

## Notas

> ⚠️ **La lista tiene que estar ordenada.** Si no lo está, la función puede devolver `-1` aunque el elemento exista.

> 💡 Si la lista cambia seguido y se busca pocas veces, ordenarla cada vez cuesta O(n log n): en ese caso puede convenir una búsqueda lineal.

<details>
<summary>▶ Errores comunes</summary>

- **Usar `medio = (izq + der) / 2`**: con `/` el resultado es `float` y no sirve como índice. Hay que usar `//`.
- **Condición `while izq < der`**: deja sin revisar el caso en que queda un solo elemento. La condición correcta es `izq <= der`.
- **Actualizar `izq = medio` en lugar de `medio + 1`**: puede generar un bucle infinito.
</details>

## Resumen de sintaxis Markdown

| Elemento | Sintaxis | Resultado |
|:---|:---|:---|
| H1 | `# Título` | Encabezado grande |
| Negrita | `**texto**` | **texto** |
| Cursiva | `*texto*` | *texto* |
| Código inline | `` `codigo` `` | `codigo` |
| Bloque código | ` ```python ` | Bloque coloreado |
| Lista | `- item` o `1. item` | Lista |
| Tarea | `- [x] tarea` | ✅ tarea |
| Tabla | `\| col \| col \|` | Tabla renderizada |
| Link | `[texto](url)` | Link clickeable |
| Imagen | `![alt](url)` | Imagen renderizada |
