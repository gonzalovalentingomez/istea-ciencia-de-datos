### 🟢 Ejercicio 1 — Clase Vector2D completa
Con __add__, __sub__, __mul__ (escalar), __repr__, __eq__, y magnitud como propiedad.


```python
import math

class Vector2D:
    def __init__(self, x, y):
        """Inicializa el vector con sus coordenadas (x, y)."""
        self.x = x
        self.y = y

    def __add__(self, other):
        """Define el comportamiento del operador de suma (+)."""
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        """Define el comportamiento del operador de resta (-)."""
        return Vector2D(self.x - other.x, self.y - other.y)

    def __mul__(self, escalar):
        """Define el comportamiento de la multiplicación (*) por un número (escalar)."""
        return Vector2D(self.x * escalar, self.y * escalar)

    def __repr__(self):
        """Define cómo se mostrará el objeto al imprimirlo, por ejemplo: (3, 4)."""
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        """Define el comportamiento del operador de igualdad (==)."""
        if isinstance(other, Vector2D):
            return self.x == other.x and self.y == other.y
        return False

    @property
    def magnitud(self):
        """
        Calcula la longitud del vector.
        El decorador @property permite acceder a esto usando `v.magnitud`
        en lugar de `v.magnitud()`.
        """
        return math.hypot(self.x, self.y) # Equivalente a math.sqrt(x**2 + y**2)

# --- Pruebas del Ejercicio 1 ---
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print("Suma:", v1 + v2)           # -> (4, 6)
print("Magnitud:", v1.magnitud)   # -> 5.0
print("Igualdad:", v1 == v1)      # -> True
```

### 🟡 Ejercicio 2 — Jerarquía Animal
Clase base Animal + subclases Perro, Gato, Pajaro, cada una con hablar() y uso de super().


```python
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def hablar(self):
        return "..."


class Perro(Animal):
    def __init__(self, nombre, edad):
        # super() llama al constructor de la clase padre (Animal)
        super().__init__(nombre, edad)

    def hablar(self):
        # Sobrescribimos el comportamiento (Polimorfismo)
        return "¡Guau, guau!"

class Gato(Animal):
    def __init__(self, nombre, edad):
        super().__init__(nombre, edad)

    def hablar(self):
        return "¡Miau!"

class Pajaro(Animal):
    def __init__(self, nombre, edad, color_plumas="Amarillo"):
        # Llamamos al padre y agregamos un atributo exclusivo de Pajaro
        # para demostrar la utilidad real de super()
        super().__init__(nombre, edad)
        self.color_plumas = color_plumas

    def hablar(self):
        return "¡Pío pío!"

# --- Pruebas del Ejercicio 2 ---
perro = Perro("Firulais", 5)
gato = Gato("Michi", 3)
pajaro = Pajaro("Tweety", 1, "Azul")

print(f"{perro.nombre} (Perro) dice: {perro.hablar()}")
print(f"{gato.nombre} (Gato) dice: {gato.hablar()}")
print(f"{pajaro.nombre} (Pajaro) dice: {pajaro.hablar()}")
print(f"{pajaro.nombre} (Pajaro) es de color: {pajaro.color_plumas}")
```

### 🟡 Ejercicio 3 — FechaPersonalizada con comparación
Con __lt__, __gt__, __eq__ para comparar fechas y __str__ en formato DD/MM/AAAA.


```python
class FechaPersonalizada:
    def __init__(self, dia, mes, año):
        self.dia = dia
        self.mes = mes
        self.año = año

    def __str__(self):
        """
        Define cómo se representa el objeto como cadena de texto (str).
        El :02d indica que debe tener 2 dígitos, rellenando con ceros a la izquierda si es necesario.
        """
        return f"{self.dia:02d}/{self.mes:02d}/{self.año}"

    def _como_tupla(self):
        """
        Método auxiliar (privado por convención) para facilitar la comparación.
        El orden (año, mes, dia) asegura que la comparación cronológica sea correcta.
        """
        return (self.año, self.mes, self.dia)

    def __lt__(self, other):
        """Define el comportamiento del operador 'menor que' (<)"""
        if isinstance(other, FechaPersonalizada):
            return self._como_tupla() < other._como_tupla()
        return NotImplemented

    def __gt__(self, other):
        """Define el comportamiento del operador 'mayor que' (>)"""
        if isinstance(other, FechaPersonalizada):
            return self._como_tupla() > other._como_tupla()
        return NotImplemented

    def __eq__(self, other):
        """Define el comportamiento del operador de 'igualdad' (==)"""
        if isinstance(other, FechaPersonalizada):
            return self._como_tupla() == other._como_tupla()
        return False

# --- Pruebas del Ejercicio 3 ---
f1 = FechaPersonalizada(1, 1, 2024)
f2 = FechaPersonalizada(15, 6, 2024)

print("f1 < f2:", f1 < f2)   # -> True
print("f1 > f2:", f1 > f2)   # -> False
print("f1 == f1:", f1 == f1) # -> True
print("f1 como texto:", str(f1))  # -> '01/01/2024'
```

### 🔴 Ejercicio 4 — Name mangling
Demostrar name mangling: clase BankAccount con __balance y mostrar acceso _BankAccount__balance.


```python
class CuentaBanco:
    def __init__(self, titular, saldo):
        self.__saldo = saldo  # name mangling: se guarda internamente como _CuentaBanco__saldo
        self.titular = titular

    def depositar(self, monto):
        self.__saldo += monto

# --- Demostración del Ejercicio 4 ---
cuenta = CuentaBanco("Ana", 5000)

# 1. Intentamos acceder al atributo de forma directa (esto fallará)
print("Intentando acceder a __saldo directamente...")
try:
    print(cuenta.__saldo)
except AttributeError as e:
    print(f"❌ Error capturado: {e}")
    print("Explicación: El atributo '__saldo' parece no existir desde afuera de la clase.\n")

# 2. Accedemos utilizando el nombre modificado por el Name Mangling (esto funcionará)
print("Accediendo mediante el nombre modificado (_CuentaBanco__saldo)...")
try:
    saldo_oculto = cuenta._CuentaBanco__saldo
    print(f"✅ Éxito: El saldo es {saldo_oculto}")
except AttributeError:
    print("Esto no debería fallar.")
```

### 🔴 Ejercicio 5 — Mixin de logging
Crear LogMixin que agrega logging a cualquier método de cualquier clase.


```python
import datetime

class LogMixin:
    def log(self, mensaje):
        """
        Imprime un mensaje de registro que incluye la fecha/hora,
        el nombre de la clase actual y el mensaje proporcionado.
        """
        # Obtenemos el timestamp actual formateado
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        # Obtenemos dinámicamente el nombre de la clase hija
        clase = self.__class__.__name__

        # Imprimimos el resultado final
        print(f"[{timestamp}] [{clase}] {mensaje}")

class Servicio(LogMixin):
    def __init__(self, nombre):
        self.nombre = nombre

    def ejecutar(self, tarea):
        # Utilizamos el método heredado del Mixin
        self.log(f"Ejecutando: {tarea}")
        return f"✓ {tarea} completada"

# --- Pruebas del Ejercicio 5 ---
s = Servicio("Servicio de Pagos")

# Al ejecutar, primero veremos el print del log dentro de ejecutar()
resultado = s.ejecutar("procesar pago")

# Y luego el resultado devuelto por la función
print(resultado)
```

## Desafío extra 🔴

Implementar un sistema de **observadores** (patrón Observer) con:
- Clase `EventoMixin` que agrega `on(evento, callback)` y `emit(evento, *args)`
- Clase `Stock` que hereda de `EventoMixin` y emite eventos cuando el stock baja de un umbral
- Al menos 3 observadores (funciones) que reaccionan al evento


```python
# 🔴 Desafío: Patrón Observer
class EventoMixin:
    """Permite suscribir funciones a eventos y avisarles cuando ocurren."""

    def on(self, evento, callback):
        self.__dict__.setdefault("_observers", {})
        self._observers.setdefault(evento, []).append(callback)

    def emit(self, evento, *args):
        for callback in self.__dict__.get("_observers", {}).get(evento, []):
            callback(*args)

class StockProducto(EventoMixin):
    def __init__(self, nombre, cantidad):
        self.nombre = nombre
        self._cantidad = cantidad
        self._observers = {}

    def vender(self, cantidad):
        self._cantidad -= cantidad
        if self._cantidad < 5:
            self.emit("stock_bajo", self.nombre, self._cantidad)

# Observadores
def alerta_consola(nombre, cantidad):
    print(f"   ⚠️  Stock bajo de {nombre}: quedan {cantidad} unidades")

historial_alertas = []
def registrar_alerta(nombre, cantidad):
    historial_alertas.append((nombre, cantidad))

def pedir_reposicion(nombre, cantidad, stock_objetivo=10):
    print(f"   📦 Pedido automático: {stock_objetivo - cantidad} unidades de {nombre}")

yerba = StockProducto("Yerba 1kg", 12)
for observador in (alerta_consola, registrar_alerta, pedir_reposicion):
    yerba.on("stock_bajo", observador)

for venta in [3, 4, 2, 1]:
    print(f"Venta de {venta} unidades")
    yerba.vender(venta)

print(f"\nHistorial de alertas: {historial_alertas}")
```

## Resumen de conceptos

| Concepto | Sintaxis | Descripción |
|---|---|---|
| @property | `@property` / `@attr.setter` | Encapsula acceso con validación |
| Name mangling | `self.__attr` → `_Clase__attr` | Pseudo-privado |
| __repr__ | `def __repr__(self):` | Representación para desarrolladores |
| __str__ | `def __str__(self):` | Representación para usuarios |
| __eq__ | `def __eq__(self, otro):` | Operador == |
| __add__ | `def __add__(self, otro):` | Operador + |
| __iter__ | `def __iter__(self):` | Hace el objeto iterable |
| @staticmethod | `@staticmethod` | Método sin self ni cls |
| @classmethod | `@classmethod` | Método que recibe cls |
| super() | `super().__init__(...)` | Llama método del padre |
| Mixin | `class M: ...` (sin estado) | Herencia múltiple modular |
