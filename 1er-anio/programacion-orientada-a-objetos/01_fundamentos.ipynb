## Ejercicios

### 🟢 Ejercicio 1 — Variables y tipos
Declarar variables de 4 tipos distintos (int, float, str, bool) y mostrar su valor y tipo con `type()`.


```python
# 🎯 EJERCICIO 1: Declarar variables de 4 tipos y mostrar su tipo
nombre = "Carlos"
edad = 25
peso = 72.5
activo = True

for etiqueta, valor in [("nombre", nombre), ("edad", edad), ("peso", peso), ("activo", activo)]:
    print(f"{etiqueta} = {valor!r}  →  {type(valor)}")
```

### 🟢 Ejercicio 2 — Operadores
Dados `x = 17` e `y = 4`, imprimir el resultado de `//`, `%` y `**`.


```python
# 🎯 EJERCICIO 2: Operadores especiales con x=17, y=4
x, y = 17, 4
print(f"{x} // {y} = {x // y}")   # división entera
print(f"{x} % {y}  = {x % y}")    # resto
print(f"{x} ** {y} = {x ** y}")   # potencia
```

### 🟡 Ejercicio 3 — Clasificación de temperatura
Dada una temperatura en Celsius, clasificarla: < 0 → helada, 0-15 → fría, 16-25 → templada, > 25 → calurosa.


```python
# 🎯 EJERCICIO 3: Clasificar temperatura
def clasificar_temperatura(t):
    if t < 0:
        return "helada"
    elif t <= 15:
        return "fría"
    elif t <= 25:
        return "templada"
    return "calurosa"

temperatura = 18
print(f"La temperatura de {temperatura}°C es {clasificar_temperatura(temperatura)}.")

# Pruebo con otros valores, incluyendo los bordes de cada rango
for t in [-3, 0, 15, 16, 25, 31]:
    print(f"  {t:>3}°C → {clasificar_temperatura(t)}")
```

### 🟡 Ejercicio 4 — Múltiplos de 7
Imprimir todos los múltiplos de 7 entre 1 y 100 usando `for` + `range`.


```python
# 🎯 EJERCICIO 4: Múltiplos de 7 entre 1 y 100
for n in range(7, 101, 7):   # arranca en 7 y avanza de a 7
    print(n, end=" ")
print()
```

### 🔴 Ejercicio 5 — FizzBuzz
Del 1 al 50: si es divisible por 3 imprimir "Fizz", por 5 imprimir "Buzz", por ambos "FizzBuzz", si no el número.


```python
# 🎯 EJERCICIO 5: FizzBuzz del 1 al 50
resultado = []
for n in range(1, 51):
    if n % 3 == 0 and n % 5 == 0:   # primero el caso de ambos
        resultado.append("FizzBuzz")
    elif n % 3 == 0:
        resultado.append("Fizz")
    elif n % 5 == 0:
        resultado.append("Buzz")
    else:
        resultado.append(str(n))

print(" ".join(resultado))
```

## Desafío extra 🔴

Escribir un programa que simule un cajero automático simplificado:
- El usuario tiene un saldo inicial de $5000
- En un bucle while, mostrar opciones: 1) Consultar saldo, 2) Extraer dinero, 3) Salir
- Si extrae más de lo que tiene, mostrar error sin modificar el saldo
- Al salir, mostrar el saldo final y la cantidad de operaciones realizadas


```python
# 🔴 Desafío: Cajero automático simplificado
operaciones_simuladas = [
    ("consultar", 0),
    ("extraer", 1500),
    ("extraer", 4000),   # debe fallar: saldo insuficiente
    ("extraer", 800),
    ("salir", 0)
]

saldo = 5000
operaciones_realizadas = 0
i = 0

while i < len(operaciones_simuladas):
    tipo, monto = operaciones_simuladas[i]
    i += 1

    if tipo == "salir":
        break

    operaciones_realizadas += 1
    if tipo == "consultar":
        print(f"Saldo actual: ${saldo}")
    elif tipo == "extraer":
        if monto > saldo:
            print(f"❌ No se pueden extraer ${monto}: saldo insuficiente (${saldo})")
        else:
            saldo -= monto
            print(f"✅ Extracción de ${monto}. Saldo restante: ${saldo}")

print(f"\nSaldo final: ${saldo} | Operaciones realizadas: {operaciones_realizadas}")
```

## Resumen de conceptos

| Concepto | Sintaxis | Descripción |
|---|---|---|
| Variable | `nombre = valor` | Asigna un valor a un nombre |
| Tipos básicos | `int`, `float`, `str`, `bool` | Tipos de datos primitivos |
| Aritmética | `+` `-` `*` `/` `//` `%` `**` | Operaciones matemáticas |
| Condicional | `if cond: ... elif: ... else:` | Ejecución condicional |
| match-case | `match var: case val:` | Switch moderno (Python 3.10+) |
| while | `while condicion:` | Bucle con condición |
| for + range | `for i in range(n):` | Bucle con iterador |
| break | `break` | Sale del bucle actual |
| continue | `continue` | Salta a la siguiente iteración |
| f-string | `f"texto {variable}"` | Interpolación de strings |
