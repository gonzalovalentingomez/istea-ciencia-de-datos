### 🟢 Ejercicio 1 — CRUD en diccionarios
Crear un dict de estudiante, agregar la clave "promedio", actualizar "edad" y eliminar "ciudad".


```python
# 🎯 EJERCICIO 1: CRUD en diccionario de estudiante
estudiante = {"nombre": "Pablo", "edad": 20, "ciudad": "Rosario"}

estudiante["promedio"] = 7.8   # Create
estudiante["edad"] = 21        # Update
del estudiante["ciudad"]       # Delete

print(estudiante)              # Read
```

### 🟡 Ejercicio 2 — Operaciones con sets de emails
Dadas dos listas de emails, encontrar: solo en lista1, solo en lista2, en ambas.


```python
# 🎯 EJERCICIO 2: Operaciones con sets de emails
lista1 = ["ana@mail.com", "carlos@mail.com", "lucia@mail.com", "martin@mail.com"]
lista2 = ["martin@mail.com", "diego@mail.com", "lucia@mail.com", "sofia@mail.com"]

set1, set2 = set(lista1), set(lista2)

print("Solo en lista1:", sorted(set1 - set2))
print("Solo en lista2:", sorted(set2 - set1))
print("En ambas:      ", sorted(set1 & set2))
```

### 🟡 Ejercicio 3 — Contador de palabras
Dada una oración, contar las ocurrencias de cada palabra usando un diccionario.


```python
# 🎯 EJERCICIO 3: Frecuencia de palabras
oracion = "el sol sale todos los dias y el cielo se tiñe de naranja todos los dias"

frecuencia = {}
for palabra in oracion.split():
    frecuencia[palabra] = frecuencia.get(palabra, 0) + 1

# Muestro de la más repetida a la menos repetida
for palabra, cantidad in sorted(frecuencia.items(), key=lambda x: x[1], reverse=True):
    print(f"  {palabra!r}: {cantidad}")
```

### 🟡 Ejercicio 4 — Dict comprehension con filtro
De un dict nombre→nota, crear un nuevo dict solo con estudiantes que tienen nota >= 7.


```python
# 🎯 EJERCICIO 4: Filtrar aprobados con dict comprehension
notas = {"Ana": 8, "Beto": 5, "Clara": 9, "Diego": 6, "Elena": 7, "Franco": 4}

aprobados = {nombre: nota for nombre, nota in notas.items() if nota >= 7}
print(aprobados)
```

### 🔴 Ejercicio 5 — Eliminar duplicados preservando orden
Dada una lista con duplicados, eliminarlos preservando el orden de primera aparición.


```python
# 🎯 EJERCICIO 5: Eliminar duplicados preservando orden
lista_con_duplicados = ["banana", "manzana", "banana", "naranja", "manzana", "uva", "naranja", "kiwi"]

# Las claves de un dict no se repiten y mantienen el orden de inserción
sin_duplicados = list(dict.fromkeys(lista_con_duplicados))
print(sin_duplicados)
```

## Desafío extra 🔴

Implementar un sistema de **índice invertido** para búsqueda de texto:
- Dado un diccionario de documentos `{doc_id: texto}`, construir un índice donde
  cada palabra mapea al conjunto de documentos que la contienen.
- Función `buscar(indice, *palabras)` que retorne los documentos que contienen TODAS las palabras.
- Probar con al menos 5 documentos y búsquedas de 2-3 palabras.


```python
# 🔴 Desafío: Índice invertido para búsqueda
import re

documentos = {
    1: "Python es un lenguaje de programación orientado a objetos",
    2: "La programación en Python permite crear aplicaciones web y de ciencia de datos",
    3: "Los objetos en Python tienen atributos y métodos",
    4: "El aprendizaje automático usa Python y sus bibliotecas como NumPy y Pandas",
    5: "Python es ideal para la automatización y el análisis de datos",
}

def construir_indice(docs):
    """Cada palabra apunta al conjunto de documentos donde aparece."""
    indice = {}
    for doc_id, texto in docs.items():
        for palabra in re.findall(r"\w+", texto.lower()):
            indice.setdefault(palabra, set()).add(doc_id)
    return indice

def buscar(indice, *palabras):
    """Devuelve los documentos que contienen TODAS las palabras (intersección)."""
    conjuntos = [indice.get(p.lower(), set()) for p in palabras]
    return set.intersection(*conjuntos) if conjuntos else set()

indice = construir_indice(documentos)
print(f"Palabras indexadas: {len(indice)}\n")

consultas = [("python", "datos"), ("objetos", "python"), ("programación", "web"),
             ("numpy", "pandas"), ("python", "java")]
for consulta in consultas:
    print(f"  {' + '.join(consulta):<22} → documentos {sorted(buscar(indice, *consulta))}")
```

## Resumen de conceptos

| Concepto | Ejemplo | Descripción |
|---|---|---|
| Dict literal | `{"k": v}` | Crear diccionario |
| Dict get | `d.get("k", default)` | Leer sin KeyError |
| Dict items | `d.items()` | Pares (clave, valor) |
| Dict comprehension | `{k: v for k, v in d.items()}` | Crear dict compacto |
| set | `{1, 2, 3}` | Conjunto sin duplicados |
| Unión | `a | b` | Todos los elementos |
| Intersección | `a & b` | Elementos comunes |
| Diferencia | `a - b` | En a pero no en b |
| Dif. simétrica | `a ^ b` | En uno pero no en ambos |
| frozenset | `frozenset({1,2,3})` | Conjunto inmutable |
