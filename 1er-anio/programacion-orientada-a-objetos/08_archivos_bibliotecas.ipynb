### 🟢 Ejercicio 1 — CSV: escribir y leer
Escribir una lista de dicts a CSV, luego leerlo e imprimirlo formateado.


```python
import csv
import os

# Datos iniciales
productos = [
    {"nombre": "Notebook", "precio": 350000, "stock": 15},
    {"nombre": "Mouse", "precio": 8500, "stock": 50},
    {"nombre": "Teclado", "precio": 15000, "stock": 30},
    {"nombre": "Monitor", "precio": 180000, "stock": 8},
]
archivo = "/tmp/ejercicio_productos.csv"

# Asegurar que el directorio /tmp exista (especialmente útil si estás en Windows)
os.makedirs(os.path.dirname(archivo), exist_ok=True)

# 1. ESCRIBIR EL ARCHIVO CSV
with open(archivo, mode="w", newline="", encoding="utf-8") as f:
    # Definimos los nombres de las columnas basándonos en las llaves del diccionario
    columnas = ["nombre", "precio", "stock"]
    writer = csv.DictWriter(f, fieldnames=columnas)

    # Escribimos la cabecera y las filas de datos
    writer.writeheader()
    writer.writerows(productos)

print(f"✅ Archivo creado con éxito en: {archivo}\n")

# 2. LEER EL ARCHIVO CSV Y MOSTRAR RESULTADO FORMATEADO
print(f"{'PRODUCTO':<15} | {'PRECIO':<10} | {'STOCK':<5}")
print("-" * 38)

with open(archivo, mode="r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for fila in reader:
        # Formateamos la salida para que quede alineada como una tabla
        nombre = fila["nombre"]
        # Convertimos a número para poder darle formato de moneda si se quisiera,
        # pero mantenemos la estética limpia que pide el ejercicio.
        precio = int(fila["precio"])
        stock = int(fila["stock"])

        print(f"{nombre:<15} | ${precio:<9} | {stock:<5}")
```

### 🟡 Ejercicio 2 — JSON: leer, modificar y guardar
Crear un archivo JSON de configuración, modificar un valor y guardarlo.


```python
# 🎯 EJERCICIO 2: Leer, modificar y guardar JSON
import json

config_inicial = {
    "nombre_app": "MiApp",
    "version": "1.0.0",
    "debug": True,
    "max_usuarios": 100
}
archivo_config = "/tmp/ejercicio_config.json"

# 1. Guardar
with open(archivo_config, "w", encoding="utf-8") as f:
    json.dump(config_inicial, f, indent=4)

# 2. Leer
with open(archivo_config, encoding="utf-8") as f:
    config = json.load(f)

# 3. Modificar
config["version"] = "1.1.0"
config["debug"] = False

# 4. Guardar de vuelta
with open(archivo_config, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=4)

# 5. Leer y mostrar el resultado final
with open(archivo_config, encoding="utf-8") as f:
    print(json.dumps(json.load(f), indent=4))
```

### 🟡 Ejercicio 3 — os.path
Verificar existencia, tamaño y extensión de un archivo.


```python
# 🎯 EJERCICIO 3: Operaciones con os.path
import os

rutas = [
    "/tmp/ejercicio_productos.csv",
    "/tmp/ejercicio_config.json",
    "/tmp/archivo_que_no_existe.txt"
]

for ruta in rutas:
    nombre = os.path.basename(ruta)
    extension = os.path.splitext(ruta)[1]
    if os.path.exists(ruta):
        print(f"✅ {nombre:<28} existe | {os.path.getsize(ruta):>4} bytes | extensión: {extension}")
    else:
        print(f"❌ {nombre:<28} no existe | extensión: {extension}")
```

### 🟡 Ejercicio 4 — datetime: clasificar fechas
De una lista de 20 fechas, encontrar cuántas son días hábiles, fines de semana, y cuál es la más reciente.


```python
# 🎯 EJERCICIO 4: Análisis de fechas
from datetime import date
import random

random.seed(42)
fechas = [date(2025, random.randint(1,12), random.randint(1,28)) for _ in range(20)]

habiles = [f for f in fechas if f.weekday() < 5]     # lunes (0) a viernes (4)
fines_de_semana = [f for f in fechas if f.weekday() >= 5]

print(f"Días hábiles: {len(habiles)} | Fines de semana: {len(fines_de_semana)}")
print(f"Fecha más reciente: {max(fechas)}")
```

### 🔴 Ejercicio 5 — re: extraer teléfonos
Extraer todos los teléfonos de un texto en formatos XXX-XXXX-XXXX y (XXX) XXXX-XXXX.


```python
# 🎯 EJERCICIO 5: Extraer teléfonos con re
import re

texto = """
Llame al 011-4433-2211 para soporte técnico.
Ventas: (351) 4567 8901 o (261) 4321-5678.
Emergencias: 0800-555-1234.
Email: info@empresa.com (no es teléfono, no debe aparecer).
"""

# Código de área de 3 o 4 dígitos (con o sin paréntesis), separador espacio o guion,
# un bloque de 3 o 4 dígitos, otro separador y un bloque final de 4 dígitos
patron = r"\(?\d{3,4}\)?[\s-]\d{3,4}[\s-]\d{4}"
telefonos = re.findall(patron, texto)
print(telefonos)
```

## Desafío extra 🔴

Implementar un **analizador de logs** que lea un archivo de log (a crear con datos sintéticos),
extraiga con regex: timestamp, nivel (INFO/WARNING/ERROR), mensaje, e IP si la hay.
Generar un reporte en JSON con: total por nivel, errores más frecuentes, IPs únicas, timeline por hora.


```python
# 🔴 Desafío: Analizador de logs
import re, json, random, os
from collections import Counter
from datetime import datetime, timedelta

TRABAJO_DIR = "/tmp/curso_python_u8"
os.makedirs(TRABAJO_DIR, exist_ok=True)
random.seed(42)

# Generar archivo de log sintético
archivo_log = os.path.join(TRABAJO_DIR, "app.log")
niveles = ["INFO"] * 6 + ["WARNING"] * 2 + ["ERROR"] * 2
mensajes_error = ["Connection timeout", "DB connection failed", "Auth token expired"]
mensajes_info = ["User logged in", "Request processed", "Cache hit", "Session started"]

with open(archivo_log, "w") as f:
    base = datetime(2026, 3, 1, 8, 0, 0)
    for i in range(100):
        ts = (base + timedelta(minutes=i*5)).strftime("%Y-%m-%d %H:%M:%S")
        nivel = random.choice(niveles)
        msg = random.choice(mensajes_error) if nivel == "ERROR" else random.choice(mensajes_info)
        ip = f"192.168.1.{random.randint(1,20)}"
        f.write(f"[{ts}] [{nivel}] {msg} from {ip}\n")

# Parsear cada línea con grupos con nombre
patron = re.compile(
    r"\[(?P<timestamp>[\d\- :]+)\] \[(?P<nivel>\w+)\] (?P<mensaje>.+?)"
    r"(?: from (?P<ip>\d{1,3}(?:\.\d{1,3}){3}))?$"
)
registros = []
with open(archivo_log) as f:
    for linea in f:
        coincidencia = patron.match(linea.strip())
        if coincidencia:
            registros.append(coincidencia.groupdict())

reporte = {
    "total_lineas": len(registros),
    "total_por_nivel": dict(Counter(r["nivel"] for r in registros)),
    "errores_mas_frecuentes": Counter(r["mensaje"] for r in registros if r["nivel"] == "ERROR").most_common(3),
    "ips_unicas": sorted({r["ip"] for r in registros if r["ip"]}, key=lambda ip: int(ip.split(".")[-1])),
    "timeline_por_hora": dict(sorted(Counter(r["timestamp"][11:13] + "h" for r in registros).items())),
}

ruta_reporte = os.path.join(TRABAJO_DIR, "reporte_logs.json")
with open(ruta_reporte, "w", encoding="utf-8") as f:
    json.dump(reporte, f, indent=2, ensure_ascii=False)

print(f"Reporte guardado en {ruta_reporte}\n")
print(json.dumps(reporte, indent=2, ensure_ascii=False))
```

## Resumen de conceptos

| Herramienta | Función clave | Uso |
|---|---|---|
| open/with | `with open(path, mode) as f:` | Archivos de texto |
| csv | `DictWriter / DictReader` | Archivos CSV |
| json | `json.dump / json.load` | Archivos JSON |
| os.path | `exists, getsize, splitext` | Sistema de archivos |
| pathlib | `Path(ruta).name, .suffix` | Rutas OO |
| datetime | `date.today(), timedelta` | Fechas y duraciones |
| random | `randint, choice, shuffle` | Aleatoriedad |
| re | `findall, match, search, sub` | Expresiones regulares |
