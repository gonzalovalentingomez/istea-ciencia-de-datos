### 🟢 Ejercicio 1 — Clase Rectangulo
Clase con ancho y alto, métodos: area(), perimetro(), es_cuadrado().


```python
# 🎯 EJERCICIO 1: Clase Rectangulo
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

    def es_cuadrado(self):
        return self.ancho == self.alto

# Pruebas:
r1 = Rectangulo(5, 3)
r2 = Rectangulo(4, 4)
print(r1.area())        # → 15
print(r1.perimetro())   # → 16
print(r1.es_cuadrado()) # → False
print(r2.es_cuadrado()) # → True
```

### 🟡 Ejercicio 2 — Clase CuentaBancaria
Con saldo, métodos depositar/extraer/saldo. No permitir saldo negativo.


```python
# 🎯 EJERCICIO 2: Clase CuentaBancaria simplificada
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self._saldo = saldo_inicial

    def depositar(self, monto):
        if monto <= 0:
            print("Error: el monto a depositar debe ser positivo")
            return
        self._saldo += monto

    def extraer(self, monto):
        if monto > self._saldo:
            print(f"Error: saldo insuficiente para extraer ${monto} (saldo: ${self._saldo})")
            return
        self._saldo -= monto

    def get_saldo(self):
        return self._saldo

# Pruebas:
cuenta = CuentaBancaria("Laura", 500)
cuenta.depositar(300)
print(cuenta.get_saldo())  # → 800
cuenta.extraer(1000)       # debe mostrar error
print(cuenta.get_saldo())  # el saldo no cambia → 800
```

### 🟡 Ejercicio 3 — isinstance y setattr
Crear 3 instancias de Estudiante, verificar con isinstance, agregar atributo "promedio" con setattr.


```python
# 🎯 EJERCICIO 3: isinstance y setattr con Estudiante
class Estudiante:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

estudiantes = [Estudiante("Ana", 20), Estudiante("Bruno", 22), Estudiante("Carla", 19)]
promedios = [8.5, 7.2, 9.1]

for estudiante, promedio in zip(estudiantes, promedios):
    print(f"¿{estudiante.nombre} es Estudiante? {isinstance(estudiante, Estudiante)}")
    setattr(estudiante, "promedio", promedio)   # agrego el atributo dinámicamente

print()
for e in estudiantes:
    print(f"{e.nombre}: {vars(e)}")
```

### 🟡 Ejercicio 4 — Atributo de clase contador
Clase Vehiculo con atributo de clase total_vehiculos que incrementa con cada instancia.


```python
# 🎯 EJERCICIO 4: Atributo de clase contador
class Vehiculo:
    total_vehiculos = 0     # atributo de clase: lo comparten todas las instancias

    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        Vehiculo.total_vehiculos += 1

# Pruebas:
v1 = Vehiculo("Toyota", "Hilux")
v2 = Vehiculo("Ford", "Ranger")
v3 = Vehiculo("Chevrolet", "S10")
print(Vehiculo.total_vehiculos)  # → 3
```

### 🔴 Ejercicio 5 — Propiedad de conversión de temperatura
Clase con propiedad celsius que convierte automáticamente a fahrenheit y kelvin.


```python
# 🎯 EJERCICIO 5: Clase con propiedad de temperatura
class ConvertidorTemperatura:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    @property
    def kelvin(self):
        return self.celsius + 273.15

# Pruebas:
t = ConvertidorTemperatura(100)
print(t.fahrenheit)  # → 212.0
print(t.kelvin)      # → 373.15

t.celsius = 0        # si cambio celsius, las propiedades se recalculan solas
print(t.fahrenheit, t.kelvin)
```

## Desafío extra 🔴

Implementar una clase `Pila` (Stack) que encapsule una lista y exponga únicamente
las operaciones válidas de una pila: `push(item)`, `pop()`, `peek()`, `esta_vacia()`, `tamaño()`.
Agregar un método `__repr__` que muestre el estado actual.
Demostrar su uso con el algoritmo de verificación de paréntesis balanceados.


```python
# 🔴 Desafío: Clase Pila y verificación de paréntesis
class Pila:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        if self.esta_vacia():
            raise IndexError("pop sobre una pila vacía")
        return self._items.pop()

    def peek(self):
        return None if self.esta_vacia() else self._items[-1]

    def esta_vacia(self):
        return len(self._items) == 0

    def tamaño(self):
        return len(self._items)

    def __repr__(self):
        return f"Pila({self._items})"

def verificar_parentesis(expresion):
    """Verifica si los paréntesis/corchetes/llaves están balanceados."""
    pares = {")": "(", "]": "[", "}": "{"}
    pila = Pila()
    for caracter in expresion:
        if caracter in "([{":
            pila.push(caracter)
        elif caracter in pares:
            if pila.esta_vacia() or pila.pop() != pares[caracter]:
                return False
    return pila.esta_vacia()

# Demostración de la pila
p = Pila()
for x in [1, 2, 3]:
    p.push(x)
print(p, "| peek:", p.peek(), "| tamaño:", p.tamaño())
p.pop()
print(p, "\n")

# Pruebas:
for expr in ["(a + b) * (c - d)", "[(1 + 2)] * {3}", "((a + b)", "{[}]"]:
    resultado = verificar_parentesis(expr)
    print(f"  {expr!r}: {'✓ balanceado' if resultado else '✗ desbalanceado'}")
```

## Resumen de conceptos

| Concepto | Sintaxis | Descripción |
|---|---|---|
| Clase | `class Nombre:` | Define un tipo |
| Constructor | `def __init__(self, ...):` | Se llama al crear objeto |
| Atributo instancia | `self.attr = valor` | Único por objeto |
| Atributo de clase | `Clase.attr = valor` | Compartido |
| Método | `def metodo(self):` | Función del objeto |
| isinstance | `isinstance(obj, Clase)` | Verificar tipo |
| hasattr | `hasattr(obj, "attr")` | Verificar atributo |
| getattr | `getattr(obj, "attr", default)` | Obtener por nombre |
| setattr | `setattr(obj, "attr", val)` | Asignar por nombre |
| vars | `vars(obj)` | Dict de atributos |
